```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSVs
train = pd.read_csv("/home/bauerste/5BaseTestrun/pretrained_model/train.csv", sep="\t")
eval_ = pd.read_csv("/home/bauerste/5BaseTestrun/pretrained_model/eval.csv", sep="\t")

# Two stacked subplots, sharing x-axis
fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# --- Train (top) ---
ax = axes[0]
ax.plot(train["step"], train["loss"], color="tab:blue", label="Train Loss", alpha=0.7)
ax.set_ylabel("Train Loss", color="tab:blue")
ax.tick_params(axis="y", labelcolor="tab:blue")

ax2 = ax.twinx()
ax2.plot(train["step"], train["acc"], color="tab:orange", label="Train Accuracy", alpha=0.7)
ax2.set_ylabel("Train Accuracy", color="tab:orange")
ax2.tick_params(axis="y", labelcolor="tab:orange")

ax.set_title("Pretraining: Train Metrics")
ax.grid(alpha=0.3)

# --- Eval (bottom) ---
ax = axes[1]
ax.plot(eval_["step"], eval_["test_loss"], "o-", color="tab:blue", label="Eval Loss")
ax.set_ylabel("Eval Loss", color="tab:blue")
ax.tick_params(axis="y", labelcolor="tab:blue")

ax2 = ax.twinx()
ax2.plot(eval_["step"], eval_["test_acc"], "s-", color="tab:orange", label="Eval Accuracy")
ax2.set_ylabel("Eval Accuracy", color="tab:orange")
ax2.tick_params(axis="y", labelcolor="tab:orange")

ax.set_xlabel("Training Step")
ax.set_title("Pretraining: Eval Metrics (every 500 steps)")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("/home/bauerste/5BaseTestrun/pretrained_model/loss_curves.png", dpi=150)
plt.show()
print("Saved to loss_curves.png")
```
