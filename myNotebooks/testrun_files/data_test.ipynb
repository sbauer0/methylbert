```python
# fastq
import gzip
from itertools import islice

fastq_path = "/data/gidb/shared/datasets/MethylBERT/DLBCL/fastq/SRR10099822_1.fastq.gz"

# Print first few reads (FASTQ format: 4 lines per read)
with gzip.open(fastq_path, "rt") as f:
    lines = list(islice(f, 40))  # 10 reads (4 lines each)

print("".join(lines))
```

```python
import pysam

bam_path = "/data/gidb/shared/datasets/MethylBERT/B_cell/bam/SRR10165874_1_bismark_bt2_pe.bam"

bam = pysam.AlignmentFile(bam_path, "rb")

for i, read in enumerate(bam.fetch(until_eof=True)):
    seq = read.query_sequence
    xm = read.get_tag("XM") if read.has_tag("XM") else None

    print(read.to_string())
    print(f"Sequence length: {len(seq) if seq else 'NA'}")
    print(f"XM length: {len(xm) if xm else 'NA'}")
    print("-" * 50)

    
    break

bam.close()
```
