```python
import pysam

bam_path = "/data/gidb/shared/datasets/MethylBERT/COLO829_BL/colo829/sup/PAU59949.d052sup4305mCG_5hmCGvHg38.bam"

aln = pysam.AlignmentFile(bam_path, "rb")
read = next(aln.fetch(until_eof=True))

print("=== Basic info ===")
print(f"Read name:         {read.query_name}")
print(f"Reference:         {read.reference_name}:{read.reference_start}")
print(f"Mapping quality:   {read.mapping_quality}")
print(f"Is unmapped:       {read.is_unmapped}")
print(f"Is secondary:      {read.is_secondary}")
print(f"Is supplementary:  {read.is_supplementary}")
print(f"Query length:      {read.query_length}")
print(f"Aligned length:    {read.query_alignment_length}")

print("\n=== Sequence (first 100 bp) ===")
print(read.query_sequence[:100] if read.query_sequence else "<no sequence>")

print("\n=== All tags ===")
for tag, value in read.get_tags():
    val_str = str(value)
    if len(val_str) > 120:
        val_str = val_str[:120] + f"... [truncated, full length {len(str(value))}]"
    print(f"  {tag}: {val_str}")

print("\n=== Methylation-relevant tags ===")
for tag in ("MM", "ML", "Mm", "Ml"):  # MM/ML are the SAMv1.6 spec; Mm/Ml are the older lowercase form
    if read.has_tag(tag):
        v = read.get_tag(tag)
        print(f"  {tag} present, type={type(v).__name__}, preview={str(v)[:120]}")
    else:
        print(f"  {tag}: not present")

aln.close()
```

```python
import pysam

bam_path = "/data/gidb/shared/datasets/MethylBERT/COLO829_BL/colo829/sup/PAU59949.d052sup4305mCG_5hmCGvHg38.bam"

aln = pysam.AlignmentFile(bam_path, "rb")

read = None
for r in aln.fetch(until_eof=True):
    if r.is_unmapped or r.is_secondary or r.is_supplementary:
        continue
    if r.query_sequence is None:
        continue
    read = r
    break

print("=== Basic info ===")
print(f"Read name:         {read.query_name}")
print(f"Reference:         {read.reference_name}:{read.reference_start}")
print(f"Mapping quality:   {read.mapping_quality}")
print(f"Query length:      {read.query_length}")
print(f"Aligned length:    {read.query_alignment_length}")

print("\n=== Sequence (first 100 bp) ===")
print(read.query_sequence[:100])

print("\n=== Methylation tags ===")
for tag in ("MM", "ML"):
    if read.has_tag(tag):
        v = read.get_tag(tag)
        print(f"  {tag} present, type={type(v).__name__}, preview={str(v)[:200]}")
    else:
        print(f"  {tag}: not present")

aln.close()
```
